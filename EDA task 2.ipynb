```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
df=pd.read_csv("profit_sales_analysis_dataset.csv")
```

```python
df
```

```python
sns.histplot(df["Sales"],bins=8,color="red")
plt.title("distribution of sales")
plt.show()
```

```python
sns.histplot(df["Profit"],bins=8,color="black")
plt.title("distribtion of profit")
plt.show()
```

```python
sns.boxplot(x=df["Sales"])
plt.title("boxplot of sales")
plt.show()
```

```python
sns.boxplot(x=df["Profit"],color="red")
plt.title("boxplot of profit")
plt.show()
```

```python
sns.scatterplot(x=df["Discount_Percent"],y=df["Profit_Margin_Percent"],data=df,color="black")
plt.title("Discount_Percent vs Profit_Margin_Percent")
plt.ylabel("Discount_Percent")
plt.xlabel("Profit_Margin_Percent")
plt.show()
```

```python
sns.barplot(x=df["Discount_Percent"],y=df["Profit_Margin_Percent"],data=df,color="purple")
plt.title("Discount_Percent vs Profit_Margin_Percent")
plt.ylabel("Discount_Percent")
plt.xlabel("Profit_Margin_Percent")
plt.show()
```

```python
numaric_column = [
    "Transaction_ID",
    "Unit_Price",
    "Quantity",
    "Discount_Percent",
    "Sales",
    "Profit",
    "Profit_Margin_Percent",
    "Cost",
]
correlation = df[numaric_column].corr()
```

```python
sns.heatmap(correlation,  
           annot=True,     
           cmap="Reds",  
           cbar=False,     
           square=True, 
           linewidth=2)
plt.title("all numerical")
plt.show()
```
