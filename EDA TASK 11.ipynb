```python
import pandas as pd
```

```python
df=pd.read_csv("Automobile_data.csv")
```

```python
df
```

```python
import numpy as np
```

```python
df=df.replace("?",np.nan)
```

```python
print(df.head())
```

```python
df.dtypes
```

```python
df.info()
```

```python
numeric_columns = ['horsepower', 'peak-rpm', 'price', 'bore', 'stroke']
```

```python
for col in numeric_columns:
    df[col]=pd.to_numeric(df[col],errors='coerce')
print(df[col].dtypes)
```

```python
df[['horsepower', 'peak-rpm', 'price', 'bore', 'stroke']].isnull().sum()
```

```python
df["bore"] = df["bore"].fillna(df["bore"].median())
```

```python
df["stroke"] = df["stroke"].fillna(df["stroke"].median())
```

```python
df["horsepower"] = df["horsepower"].fillna(df["horsepower"].median())
```

```python
df["peak-rpm"] = df["peak-rpm"].fillna(df["peak-rpm"].median())
```

```python
df["price"] = df["price"].fillna(df["price"].median())
```

```python
df["power_to_weight_ratio"] = df["horsepower"] / df["curb-weight"]
print(df[["horsepower","curb-weight","power_to_weight_ratio"]].head())
```

```python
df["combined_mpg"] = (df["city-mpg"] + df["highway-mpg"]) / 2
print (df[["city-mpg","highway-mpg","combined_mpg"]].head())
```

```python
body_style_summary = df.groupby('body-style')[
    ['price', 'horsepower', 'curb-weight', 'combined_mpg', 'power_to_weight_ratio']
].agg(['count', 'mean', 'median', 'max'])

print(body_style_summary)
```

```python
engine_location_summary = df.groupby('engine-location')[
    ['price', 'horsepower', 'curb-weight', 'combined_mpg', 'power_to_weight_ratio']
].agg(['count', 'mean', 'median', 'max'])

print(engine_location_summary)
```
